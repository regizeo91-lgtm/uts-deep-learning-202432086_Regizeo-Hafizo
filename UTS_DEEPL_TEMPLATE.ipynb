  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>

```python
#202432086_regizeo_hafizo
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler

# Membaca file CSV
df = pd.read_csv('melbourne.csv', on_bad_lines='skip')

# Mengambil kolom target dan membersihkan karakter non-numerik (seperti '?')
data_raw = df.iloc[:, 1].values
data_cleaned = pd.to_numeric(data_raw, errors='coerce')

# Membuat DataFrame sementara untuk membuang nilai NaN (kosong)
df_clean = pd.DataFrame({'Temp': data_cleaned}).dropna()
data = df_clean.values.astype(float)

# Normalisasi data ke rentang [0, 1]
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(data)

# Fungsi pembentukan sliding window
def create_sequences(dataset, time_step=30):
    X, y = [], []
    for i in range(len(dataset) - time_step - 1):
        a = dataset[i:(i + time_step), 0]
        X.append(a)
        y.append(dataset[i + time_step, 0])
    return np.array(X), np.array(y)

time_step = 30
X, y = create_sequences(scaled_data, time_step)

# Konversi ke PyTorch Tensor
X_tensor = torch.tensor(X, dtype=torch.float32).unsqueeze(-1)
y_tensor = torch.tensor(y, dtype=torch.float32).unsqueeze(-1)

print("Pra-pemrosesan data berhasil! Bentuk X_tensor:", X_tensor.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432086_regizeo_hafizo
import torch.nn as nn

class SequenceModel(nn.Module):
    def __init__(self, input_dim=1, hidden_dim=50, output_dim=1, num_layers=1):
        super(SequenceModel, self).__init__()
        # Mendefinisikan lapisan LSTM
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first=True)
        # Lapisan Fully Connected (Linear) untuk menghasilkan output prediksi
        self.fc = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        # Meneruskan input ke lapisan LSTM
        out, _ = self.lstm(x)
        # Mengambil output dari time step terakhir
        out = self.fc(out[:, -1, :])
        out = out.reshape(-1, 1)
        return out

# Inisialisasi model
model = SequenceModel()
print("Arsitektur Model LSTM berhasil dibuat:")
print(model)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432086_regizeo_hafizo
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# 1. Inisiasi hyperparameter (epochs, learning_rate, dll.)
epochs = 50
learning_rate = 0.01

# 2. Inisiasi fungsi Loss (MSELoss) dan Optimizer (Adam)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Simpan riwayat nilai loss pada list untuk divisualisasikan
train_losses = []
dataset = TensorDataset(X_tensor, y_tensor)
dataloader = DataLoader(dataset, batch_size=32, shuffle=False)

# 4. Jalankan proses perulangan training (Training Loop)
model.train()
for epoch in range(epochs):
    epoch_loss = 0
    for batch_X, batch_y in dataloader:
        optimizer.zero_grad()
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(dataloader)
    train_losses.append(avg_loss)

    # Menampilkan progress setiap 10 epoch
    if (epoch + 1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {avg_loss:.6f}')

print("Proses pelatihan model selesai!")
```
